```python

%pip install requests pandas openpyxl

import requests
import pandas as pd
import re
import time

from urllib.parse import quote
from requests.adapters import HTTPAdapter
from urllib3.util.retry import Retry
from openpyxl import load_workbook
from openpyxl.styles import Font, PatternFill, Alignment
from openpyxl.utils import get_column_letter

print("All imports successful!")
print("Requests version:", requests.__version__)
```

```python
# ============================================================
# 1. CONFIGURATION
# ============================================================

UNIPROT_SEARCH_URL = "https://rest.uniprot.org/uniprotkb/search"
UNIPROT_ENTRY_URL = "https://rest.uniprot.org/uniprotkb"
PDB_SEARCH_URL = "https://search.rcsb.org/rcsbsearch/v2/query"
PDB_DATA_URL = "https://data.rcsb.org/rest/v1/core"

OUTPUT_FILE = "PDB_Structure_Collection.xlsx"

# Number of PDB entity search results requested per batch.
PAGE_SIZE = 1000

# Short pause between requests.
REQUEST_DELAY = 0.05

session = requests.Session()
session.headers.update({
    "User-Agent": "ProteinStructureCollector/1.0",
    "Accept": "application/json"
})

retry = Retry(
    total=4,
    backoff_factor=0.5,
    status_forcelist=[429, 500, 502, 503, 504],
    allowed_methods=["GET", "POST"]
)

session.mount("https://", HTTPAdapter(max_retries=retry))
```

```python
# ============================================================
# 2. HELPER FUNCTIONS
# ============================================================

def get_json(url, params=None, timeout=45):
    """GET JSON with error handling."""
    response = session.get(url, params=params, timeout=timeout)
    response.raise_for_status()
    return response.json()


def normalize_name(value):
    """Normalize names for reliable text comparison."""
    if not value:
        return ""

    value = str(value).lower().strip()
    value = value.replace("β", "beta").replace("α", "alpha")
    value = re.sub(r"[^a-z0-9]+", " ", value)
    return re.sub(r"\s+", " ", value).strip()


def unique_strings(values):
    """Remove duplicates without losing original spelling."""
    output = []
    seen = set()

    for value in values:
        if not isinstance(value, str):
            continue

        value = value.strip()
        key = normalize_name(value)

        if key and key not in seen:
            seen.add(key)
            output.append(value)

    return output


def extract_uniprot_names(entry):
    """Extract recommended, alternative and gene names."""
    names = []

    description = entry.get("proteinDescription", {})

    recommended = description.get("recommendedName", {})
    full_name = recommended.get("fullName", {}).get("value")

    if full_name:
        names.append(full_name)

    for alternative in description.get("alternativeNames", []):
        full_name = alternative.get("fullName", {}).get("value")

        if full_name:
            names.append(full_name)

        for short_name in alternative.get("shortNames", []):
            if short_name.get("value"):
                names.append(short_name["value"])

    # Include recommended short names.
    for short_name in recommended.get("shortNames", []):
        if short_name.get("value"):
            names.append(short_name["value"])

    # Include gene symbols and gene synonyms.
    for gene in entry.get("genes", []):
        if gene.get("geneName", {}).get("value"):
            names.append(gene["geneName"]["value"])

        for synonym in gene.get("synonyms", []):
            if synonym.get("value"):
                names.append(synonym["value"])

        for ordered_locus in gene.get("orderedLocusNames", []):
            if ordered_locus.get("value"):
                names.append(ordered_locus["value"])

    return unique_strings(names)
```

```python
# ============================================================
# 3. IDENTIFY THE TARGET IN UNIPROT
# ============================================================

def resolve_uniprot(protein_name):
    """
    Find the most plausible UniProt entry for the input.
    Prefer an exact gene-symbol match, then an exact
    recommended protein-name match.
    """

    params = {
        "query": protein_name,
        "format": "json",
        "size": 100,
        "fields": (
            "accession,protein_name,gene_names,"
            "organism_name,reviewed"
        )
    }

    data = get_json(UNIPROT_SEARCH_URL, params=params)
    results = data.get("results", [])

    if not results:
        raise ValueError(
            f"No UniProt entries found for '{protein_name}'. "
            "Check the protein name or enter a gene symbol."
        )

    query_norm = normalize_name(protein_name)

    def rank(entry):
        genes = [
            normalize_name(g.get("geneName", {}).get("value", ""))
            for g in entry.get("genes", [])
        ]

        all_gene_names = [
            normalize_name(s.get("value", ""))
            for g in entry.get("genes", [])
            for s in g.get("synonyms", [])
        ]

        recommended = normalize_name(
            entry.get("proteinDescription", {})
            .get("recommendedName", {})
            .get("fullName", {})
            .get("value", "")
        )

        # Exact gene symbol is the strongest name-based match.
        if query_norm in genes:
            return 0

        if query_norm in all_gene_names:
            return 1

        if query_norm == recommended:
            return 2

        if query_norm in recommended:
            return 3

        # Prefer reviewed entries when otherwise comparable.
        return 4

    results.sort(key=rank)

    # If the query does not match any gene or protein name
    # exactly, warn rather than silently asserting identity.
    best = results[0]
    best_rank = rank(best)

    if best_rank == 4:
        print(
            "\nWARNING: UniProt did not return an exact gene or "
            "recommended-name match. Verify the selected target."
        )

    accession = best["primaryAccession"]

    entry = get_json(
        f"{UNIPROT_ENTRY_URL}/{quote(accession, safe='')}.json"
    )

    aliases = unique_strings(
        [protein_name] + extract_uniprot_names(entry)
    )

    print("\nTARGET IDENTIFICATION")
    print("-" * 60)
    print("Input name       :", protein_name)
    print("UniProt accession:", accession)
    print("Recommended name :", (
        entry.get("proteinDescription", {})
        .get("recommendedName", {})
        .get("fullName", {})
        .get("value", "N/A")
    ))
    print("Names to search  :")

    for name in aliases:
        print("  -", name)

    return accession, aliases
```

```python
# ============================================================
# 4. SEARCH RCSB PDB FOR EACH ALIAS
# ============================================================


def search_pdb_entities(alias):
    """Search polymer entities using the complete name phrase."""

    query = {
        "query": {
            "type": "terminal",
            "service": "text",
            "parameters": {
                "attribute": "rcsb_polymer_entity.pdbx_description",
                "operator": "contains_phrase",
                "value": alias
            }
        },
        "return_type": "polymer_entity",
        "request_options": {
            "paginate": {
                "start": 0,
                "rows": PAGE_SIZE
            },
            "results_content_type": ["experimental"]
        }
    }

    identifiers = set()
    start = 0

    while True:
        query["request_options"]["paginate"]["start"] = start

        response = session.post(
            PDB_SEARCH_URL,
            json=query,
            timeout=60
        )
        response.raise_for_status()
        result = response.json()

        batch = [
            item["identifier"]
            for item in result.get("result_set", [])
            if item.get("identifier")
        ]

        identifiers.update(batch)

        total = result.get("total_count", 0)

        if not batch or start + len(batch) >= total:
            break

        start += len(batch)
        time.sleep(REQUEST_DELAY)

    return sorted(identifiers)

   
```

```python
# ============================================================
# 5. FETCH PDB ENTITY AND ENTRY METADATA
# ============================================================

def get_polymer_entity(pdb_id, entity_id):
    return get_json(
        f"{PDB_DATA_URL}/polymer_entity/"
        f"{pdb_id}/{entity_id}"
    )


def get_pdb_entry(pdb_id):
    return get_json(
        f"{PDB_DATA_URL}/entry/{pdb_id}"
    )


# ============================================================
# 6. EXTRACT ORGANISM AND UNIPROT ACCESSIONS
# ============================================================

def get_organisms(entity):
    organisms = []

    # Naturally occurring source organism.
    for item in entity.get("rcsb_entity_source_organism", []) or []:
        name = item.get("ncbi_scientific_name")
        if name:
            organisms.append(name)

    # Recombinant expression source organism.
    for item in entity.get("rcsb_entity_host_organism", []) or []:
        name = item.get("ncbi_scientific_name")
        if name:
            organisms.append(name)

    # Some records may expose organism data under entity_src_gen.
    for item in entity.get("entity_src_gen", []) or []:
        name = item.get("pdbx_gene_src_scientific_name")
        if name:
            organisms.append(name)

    organisms = unique_strings(organisms)

    return "; ".join(organisms) if organisms else "N/A"


def get_uniprot_accessions(entity):
    ids = (
        entity.get("rcsb_polymer_entity_container_identifiers", {})
        .get("uniprot_ids", [])
    )

    ids = unique_strings(ids or [])

    return "; ".join(ids) if ids else "N/A"


# ============================================================
# 7. FALSE-POSITIVE FILTER
# ============================================================

def matches_target(description, aliases, target_accession, entity):
    """
    Accept a candidate only if:
      1. Its macromolecule description matches a target alias,
         or
      2. Its UniProt cross-reference matches the resolved
         target accession.

    Description matching uses whole normalized phrases to
    avoid accepting unrelated names such as CDK inhibitor
    proteins merely because they contain 'CDK'.
    """

    desc_norm = normalize_name(description)

    if not desc_norm:
        return False

    # A UniProt accession match is strong evidence of identity.
    uniprot_ids = (
        entity.get("rcsb_polymer_entity_container_identifiers", {})
        .get("uniprot_ids", [])
    ) or []

    if target_accession in uniprot_ids:
        return True

    alias_norms = {
        normalize_name(alias)
        for alias in aliases
        if normalize_name(alias)
    }

    for alias in alias_norms:
        # Exact description match.
        if desc_norm == alias:
            return True

        # Permit a longer construct description, e.g. a protein
        # name followed by a domain or construct description.
        # Do not use arbitrary substring matching.
        if (
            len(alias.split()) >= 2
            and (
                desc_norm.startswith(alias + " ")
                or desc_norm.endswith(" " + alias)
                or (" " + alias + " ") in (" " + desc_norm + " ")
            )
        ):
            return True

    return False

```

```python
# ============================================================
# 8. COLLECT ALL MATCHING EXPERIMENTAL STRUCTURES
# ============================================================

def collect_pdb_structures(protein_name):
    target_accession, aliases = resolve_uniprot(protein_name)

    print("\nSearching RCSB PDB...")
    entity_ids = set()

    for alias in aliases:
        try:
            found = search_pdb_entities(alias)
            entity_ids.update(found)
            print(f"{alias}: {len(found)} entity result(s)")
        except requests.RequestException as exc:
            print(f"Search warning for '{alias}': {exc}")

        time.sleep(REQUEST_DELAY)

    print("\nUnique candidate entities:", len(entity_ids))

    records = []
    processed_entries = set()

    for index, entity_identifier in enumerate(sorted(entity_ids), 1):
        try:
            pdb_id, entity_id = entity_identifier.split("_", 1)
            pdb_id = pdb_id.upper()

            entity = get_polymer_entity(pdb_id, entity_id)

            description = entity.get(
                "rcsb_polymer_entity", {}
            ).get("pdbx_description", "")

            if not matches_target(
                description, aliases, target_accession, entity
            ):
                continue

            # One output row per PDB entry.
            # If multiple matching target entities occur in
            # the same entry, combine their UniProt IDs.
            if pdb_id in processed_entries:
                existing = next(
                    r for r in records if r["PDB ID"] == pdb_id
                )

                new_ids = get_uniprot_accessions(entity)

                if new_ids != "N/A":
                    current = existing["UniProt Accession"]
                    values = (
                        [] if current == "N/A" else current.split("; ")
                    )
                    values.extend(new_ids.split("; "))
                    existing["UniProt Accession"] = "; ".join(
                        unique_strings(values)
                    )

                continue

            entry = get_pdb_entry(pdb_id)

            # Exclude theoretical/predicted entries.
            methods = entry.get("exptl", []) or []
            method_names = unique_strings([
                item.get("method", "")
                for item in methods
                if item.get("method")
            ])

            if not method_names:
                continue

            method_text = "; ".join(method_names)

            # Keep experimental methods, including X-ray,
            # electron microscopy and NMR. Exclude entries whose
            # only method is theoretical modelling.
            if any(
                "theoretical model" in m.lower()
                for m in method_names
            ):
                continue

            records.append({
                "Protein Name": protein_name,
                "PDB ID": pdb_id,
                "PDB Link": f"https://www.rcsb.org/structure/{pdb_id}",
                "Source Organism": get_organisms(entity),
                "UniProt Accession": get_uniprot_accessions(entity),
                "Method": method_text
            })

            processed_entries.add(pdb_id)

        except (requests.RequestException, ValueError, KeyError) as exc:
            print(
                f"Skipping {entity_identifier}: {exc}"
            )

        if index % 100 == 0:
            print(f"Processed {index}/{len(entity_ids)} entities")

        time.sleep(REQUEST_DELAY)

    columns = [
        "Protein Name",
        "PDB ID",
        "PDB Link",
        "Source Organism",
        "UniProt Accession",
        "Method"
    ]

    df = pd.DataFrame(records, columns=columns)

    if not df.empty:
        df = df.drop_duplicates(subset=["PDB ID"])
        df = df.sort_values("PDB ID").reset_index(drop=True)

    return df


# ============================================================
# 9. EXPORT TO FORMATTED EXCEL
# ============================================================

def save_to_excel(df, protein_name):
    safe_name = re.sub(
        r'[\\/*?:"<>|]+', "_", protein_name
    ).strip()

    output_file = f"{safe_name}_{OUTPUT_FILE}"

    with pd.ExcelWriter(output_file, engine="openpyxl") as writer:
        df.to_excel(writer, index=False, sheet_name="PDB_Structures")

    wb = load_workbook(output_file)
    ws = wb["PDB_Structures"]

    header_fill = PatternFill(
        fill_type="solid",
        fgColor="1F4E78"
    )

    for cell in ws[1]:
        cell.font = Font(bold=True, color="FFFFFF")
        cell.fill = header_fill
        cell.alignment = Alignment(
            horizontal="center",
            vertical="center",
            wrap_text=True
        )

    # Clickable PDB links.
    link_col = df.columns.get_loc("PDB Link") + 1

    for row in range(2, ws.max_row + 1):
        cell = ws.cell(row=row, column=link_col)
        if cell.value:
            cell.hyperlink = cell.value
            cell.style = "Hyperlink"

    # Readable column widths.
    widths = {
        "A": 24,
        "B": 12,
        "C": 48,
        "D": 32,
        "E": 28,
        "F": 30
    }

    for column, width in widths.items():
        ws.column_dimensions[column].width = width

    ws.freeze_panes = "A2"
    ws.auto_filter.ref = ws.dimensions
    ws.row_dimensions[1].height = 32

    wb.save(output_file)

    print("\n" + "=" * 60)
    print("COLLECTION COMPLETE")
    print("=" * 60)
    print("Protein       :", protein_name)
    print("Structures    :", len(df))
    print("Excel file    :", output_file)

    return output_file


# ============================================================
# 10. RUN: ENTER ANY TARGET PROTEIN NAME
# ============================================================

if __name__ == "__main__":
    protein_name = input(
        "Enter target protein name or gene symbol (e.g., CDK4): "
    ).strip()

    if not protein_name:
        raise ValueError("Protein name cannot be empty.")

    results_df = collect_pdb_structures(protein_name)

    if results_df.empty:
        print(
            "\nNo structures passed the target-identity filters. "
            "This does not necessarily mean that no structures exist; "
            "the aliases or matching rules may need review."
        )
    else:
        output_path = save_to_excel(results_df, protein_name)
        print("\nPreview:")
        print(results_df.head(10).to_string(index=False))
```
